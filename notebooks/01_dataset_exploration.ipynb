````python
from pathlib import Path
import json
import pandas as pd

repo = Path.cwd().resolve()
data_dir = repo / "data" / "raw" / "synthetic"
episodes = sorted(data_dir.glob("episode_*.json"))
len(episodes), episodes[:3]
````

````python
rows = []
for path in episodes:
    episode = json.loads(path.read_text())
    metadata = episode["metadata"]
    rows.append({
        "episode": path.name,
        "steps": metadata["steps"],
        "success": metadata["success"],
        "camera_queries": metadata["camera_queries"],
        "slip_events": metadata["slip_events"],
        "mean_reward": sum(episode["rewards"]) / len(episode["rewards"]),
    })

summary = pd.DataFrame(rows)
summary.head()
````

````python
summary.describe(include="all")
````

````python
if not summary.empty:
    ax = summary[["mean_reward", "camera_queries", "slip_events"]].plot(subplots=True, figsize=(10, 7), title="Episode diagnostics")
    ax
````
